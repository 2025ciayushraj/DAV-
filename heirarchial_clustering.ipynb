```python

# HIERARCHICAL CLUSTERING

# STEP 1: Import libraries

import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage

# STEP 2: Load dataset

data = load_breast_cancer()

df = pd.DataFrame(data.data, columns=data.feature_names)

print("First 5 rows of dataset:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

# STEP 3: Select features

X = df.iloc[:, :2]

# STEP 4: Standardize the data

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

# STEP 5: Create Dendrogram

linked = linkage(X_scaled[:50], method='ward')

plt.figure(figsize=(12, 6))

dendrogram(linked)

plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Data Points")
plt.ylabel("Distance")

plt.show()

# STEP 6: Apply Hierarchical Clustering

hc = AgglomerativeClustering(
    n_clusters=2,
    linkage='ward'
)

y_hc = hc.fit_predict(X_scaled)

# STEP 7: Add cluster labels

df['Cluster'] = y_hc

print("\nCluster Labels:")
print(df[['Cluster']].head(10))

# STEP 8: Visualize clusters

plt.figure(figsize=(8, 5))

plt.scatter(
    X_scaled[:, 0],
    X_scaled[:, 1],
    c=y_hc,
    cmap='rainbow'
)

plt.title("Hierarchical Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")

plt.show()

# STEP 9: Count clusters

print("\nNumber of data points in each cluster:")
print(df['Cluster'].value_counts())

# HIERARCHICAL CLUSTERING

# STEP 1: Import libraries

import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage

# STEP 2: Load dataset

data = load_breast_cancer()

df = pd.DataFrame(data.data, columns=data.feature_names)

print("First 5 rows of dataset:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

# STEP 3: Select features

X = df.iloc[:, :2]

# STEP 4: Standardize the data

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

# STEP 5: Create Dendrogram

linked = linkage(X_scaled[:50], method='ward')

plt.figure(figsize=(12, 6))

dendrogram(linked)

plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Data Points")
plt.ylabel("Distance")

plt.show()

# STEP 6: Apply Hierarchical Clustering

hc = AgglomerativeClustering(
    n_clusters=2,
    linkage='ward'
)

y_hc = hc.fit_predict(X_scaled)

# STEP 7: Add cluster labels

df['Cluster'] = y_hc

print("\nCluster Labels:")
print(df[['Cluster']].head(10))

# STEP 8: Visualize clusters

plt.figure(figsize=(8, 5))

plt.scatter(
    X_scaled[:, 0],
    X_scaled[:, 1],
    c=y_hc,
    cmap='rainbow'
)

plt.title("Hierarchical Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")

plt.show()

# STEP 9: Count clusters

print("\nNumber of data points in each cluster:")
print(df['Cluster'].value_counts())
```
