```python

# ============================================================
# FEATURE ENGINEERING: NORMALIZATION, STANDARDIZATION AND PCA
# ============================================================

# Step 1: Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.decomposition import PCA


# ============================================================
# STEP 2: CREATE A SAMPLE DATASET
# ============================================================

data = {
    'Study_Hours': [2, 4, 6, 8, 10, 3, 5, 7, 9, 12],
    'Attendance': [60, 65, 70, 75, 80, 62, 68, 78, 85, 95],
    'Assignment_Score': [40, 50, 55, 65, 70, 45, 60, 72, 80, 90],
    'Exam_Score': [45, 50, 60, 65, 75, 48, 58, 70, 82, 95]
}

df = pd.DataFrame(data)

print("ORIGINAL DATASET:")
display(df)

print("\nSTATISTICAL SUMMARY:")
display(df.describe())


# ============================================================
# STEP 3: NORMALIZATION USING MINMAXSCALER
# ============================================================

# Scale each feature between 0 and 1
minmax_scaler = MinMaxScaler()

normalized_data = minmax_scaler.fit_transform(df)

normalized_df = pd.DataFrame(
    normalized_data,
    columns=df.columns
)

print("\nNORMALIZED DATASET:")
display(normalized_df.round(2))

print("\nMinimum values:")
print(normalized_df.min().round(2))

print("\nMaximum values:")
print(normalized_df.max().round(2))


# Visualize normalized data
normalized_df.plot(
    kind='box',
    figsize=(9, 5),
    title='Normalized Features'
)

plt.ylabel("Normalized Values")
plt.grid(True)
plt.show()


# ============================================================
# STEP 4: STANDARDIZATION USING STANDARDSCALER
# ============================================================

standard_scaler = StandardScaler()

standardized_data = standard_scaler.fit_transform(df)

standardized_df = pd.DataFrame(
    standardized_data,
    columns=df.columns
)

print("\nSTANDARDIZED DATASET:")
display(standardized_df.round(2))

print("\nMean of each feature:")
print(standardized_df.mean().round(2))

print("\nStandard deviation of each feature:")
print(standardized_df.std(ddof=0).round(2))


# Visualize standardized data
standardized_df.plot(
    kind='box',
    figsize=(9, 5),
    title='Standardized Features'
)

plt.ylabel("Standardized Values")
plt.grid(True)
plt.show()


# ============================================================
# STEP 5: PRINCIPAL COMPONENT ANALYSIS (PCA)
# ============================================================

# Standardize the original dataset before PCA
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)

# Reduce 4 features to 2 principal components
pca = PCA(n_components=2)

pca_data = pca.fit_transform(scaled_data)

pca_df = pd.DataFrame(
    pca_data,
    columns=['PC1', 'PC2']
)

print("\nPCA TRANSFORMED DATA:")
display(pca_df.round(2))


# ============================================================
# STEP 6: EXPLAINED VARIANCE
# ============================================================

print("\nExplained Variance Ratio:")
print(pca.explained_variance_ratio_.round(4))

print("\nVariance retained by PC1:")
print(round(pca.explained_variance_ratio_[0] * 100, 2), "%")

print("\nVariance retained by PC2:")
print(round(pca.explained_variance_ratio_[1] * 100, 2), "%")

total_variance = pca.explained_variance_ratio_.sum()

print("\nTotal Variance Retained:")
print(round(total_variance * 100, 2), "%")


# ============================================================
# STEP 7: VISUALIZE PCA OUTPUT
# ============================================================

plt.figure(figsize=(8, 5))

plt.scatter(
    pca_df['PC1'],
    pca_df['PC2'],
    s=60
)

# Label each observation
for i in range(len(pca_df)):
    plt.annotate(
        str(i + 1),
        (pca_df['PC1'][i], pca_df['PC2'][i]),
        xytext=(5, 5),
        textcoords='offset points'
    )

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA: Dimensionality Reduction")
plt.grid(True)
plt.show()


# ============================================================
# STEP 8: FINAL COMPARISON
# ============================================================

print("\nOriginal dataset shape:", df.shape)
print("Normalized dataset shape:", normalized_df.shape)
print("Standardized dataset shape:", standardized_df.shape)
print("PCA dataset shape:", pca_df.shape)

print("\nFeature Engineering completed successfully!")
```
